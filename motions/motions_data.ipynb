```python
calico_tabs = set()

for url in urls.split('\n'): 
    if 'calicotab' not in url:
        print(f'[WARN] Skipping non-calico tab: {url}') #FIXME: could still parse some of the self-hosted ones
        continue

    calico_pattern = re.compile(r'https?:\/\/([a-zA-Z0-9-_]+)\.calicotab\.com\/([a-zA-Z0-9-_]+)\/')
    match = re.match(calico_pattern, url)

    if match is None:
        print(f'[WARN] Could not parse url: {url}')
        continue

    subdomain = match.group(1)
    slug = match.group(2)

    calico_tabs.add((subdomain, slug))

calico_tabs = list(calico_tabs)
```

```python
allowed_sides = ["og", "oo", "cg", "co"] #for now, only supports BP ballots

async def retrieve_results(pairings_url: str, session):
    """
    Retrieves the results (including ballot and speaker scores, if available) from a given pairings URL, as defined by TabbyCat.
    """

    results = {}

    pairings = None
    try:
        async with session.get(pairings_url) as pairings_response:
            if pairings_response.status == 401:
                print(f'[WARN] Behind auth for URL {pairings_url}. Skipping extracting ballots...')
                return {}

            assert pairings_response.status == 200, f"Failed to fetch {pairings_url}: {pairings_response.status}"
            pairings = await pairings_response.json()
            results["n"] = len(pairings)
    except Exception as e:
        print(f'Failed to fetch pairings for URL: {pairings_url}, {e}')
        return {}

    for pairing in pairings:
        ballots_url = pairing["_links"]["ballots"]

        try:
            ballots = None

            async with session.get(ballots_url) as ballots_response:
                assert ballots_response.status == 200, f"Failed to fetch {ballots_url}: {ballots_response.status}"
                ballots = await ballots_response.json()

            if (
                "detail" in ballots
                and ballots["detail"]
                == "Authentication credentials were not provided."
            ):
                print(
                    f"[WARN] Behind auth for URL {pairings_url}. Skipping extracting ballots..."
                )

                raise Exception(f'Failed to fetch ballots for pairing URL due to authentication error (this should not happen, since we already checked before): {ballots_url}')

            #FIXME: deal with motion used in multiple rounds
            assert len(ballots) == 1, "Expected 1 ballot for pairing URL: " + ballots_url
            sheets = ballots[0]["result"]["sheets"]

            if len(sheets) > 1:
                print(
                    f"[WARN] Multiple sheets at URL: {ballots_url}. Using only the first sheet." #FIXME:
                )

            teams = sheets[0]["teams"]

            if teams[0]["side"] not in allowed_sides:
                print(
                    f'[WARN] First team in sheet is not a valid BP position: {ballots_url}. Skipping extracting ballots...'
                )

                return {}

            for i, team in enumerate(teams):
                side = team["side"]
                assert side in allowed_sides

                points = team["points"]
                win = team["win"] 

                assert (
                    points is not None or win is not None
                ), f"Points and wins are both None for side: {side}"

                if points is not None:
                    prev_sum = results.get("sum_" + side, 0)
                    results["sum_" + side] = prev_sum + points
                
                    # sum up the points for all pairs of teams
                    # needed for balance calculation
                    for team2 in teams[i:]:
                        side2 = team2['side']

                        key = "sum_" + side + "_" + side2
                        prev_sum_sum = results.get(key, 0)
                        results[key] = prev_sum_sum + team2['points'] * points

                if win is True:
                    key = side + "_wins"
                    prev_wins = results.get(key, 0)
                    results[key] = prev_wins + 1
                
                if team["score"] is not None: 
                    key = "speaks"
                    prev_speaks = results.get(key, 0)
                    results[key] = prev_speaks + team["score"]

        except Exception as e:
            print(f'[WARN] Failed to fetch ballot for URL: {ballots_url}, {e}')
            continue

    return results
```

```python
supported_sides = ["og", "oo", "cg", "co"]

async def are_ballots_accessible(session, main_url):
    """
        Run a preliminary check to see if ballots are accessible for a given tournament (or if e.g. behind auth).
    """

    #FIXME: should check for more than just round 1, since apparently one round can be authed and the other not...
    round_1_pairings_url = main_url + "/rounds/1/pairings"

    try: 
        async with session.get(round_1_pairings_url) as round_1_pairings_response:
            if round_1_pairings_response.status == 401:
                print(f'[INFO] Behind auth for URL {round_1_pairings_url}. Skipping extracting ballots...')
                return False

            assert round_1_pairings_response.status == 200, f"Failed to fetch {round_1_pairings_url}: {round_1_pairings_response.status}"
            round_1_pairings = await round_1_pairings_response.json()

        first_pairing = round_1_pairings[0]

        ballot_url = first_pairing["_links"]["ballots"]
        async with session.get(ballot_url) as ballot_response:
            if ballot_response.status == 401:
                print(f'[INFO] Behind auth for URL {ballot_url}. Skipping extracting ballots...')
                return False
            
            assert ballot_response.status == 200, f"Failed to fetch {ballot_url}: {ballot_response.status}"

            ballots = await ballot_response.json()

            if (
                "detail" in ballots
                and ballots["detail"]
                == "Authentication credentials were not provided."
            ):
                return False
            

            sheet = ballots[0]['result']['sheets'][0]

            for team in sheet['teams']:
                if team['side'] not in supported_sides:
                    print(f'[INFO] Skipping {main_url} because it has a side that is not supported: {team["side"]}')
                    return False #FIXME: when support for wsdc/australs is added
    except Exception as e:
        print(f'Failed to check if ballots are accessible for {round_1_pairings_url}: {e}. Defaulting to True to always try to fetch ballots.')
        return True

    return True
```

```python
import asyncio 

async def retrieve_motions(subdomain, slug, session):
    rows = []

    tab_url = f"https://{subdomain}.calicotab.com/{slug}"
    
    main_url = f"https://{subdomain}.calicotab.com/api/v1/tournaments/{slug}"

    tournament_name = None
    tournament_short_name = None
    try: 
        async with session.get(main_url) as response:
            assert response.status == 200, f"Failed to fetch {main_url}: {response.status}"
            tournament_json = await response.json()

            if 'detail' in tournament_json and tournament_json['detail'] == 'No Tournament matches the given query.':
                raise Exception(f'No tournament matches the given query for URL: {main_url}')

            tournament_name = tournament_json['name']
            tournament_short_name = tournament_json['short_name']
    except Exception as e:
        print(f'[ERROR] Failed to fetch tournament root info for url: {main_url}')
        return []

    is_ballotable = await are_ballots_accessible(session, main_url)

    motions_url = f"{main_url}/motions"

    motions = None
    try: 
        async with session.get(motions_url) as response:
            assert response.status == 200, f"Failed to fetch motions for {motions_url}: {response.status}"
            motions = await response.json()
    except Exception as e:
        print(f'[ERROR] Failed to fetch motions for URL: {motions_url}, {e}')
        return []

    for motion in motions:
        if len(motion['rounds']) <= 0: #skip unused (test) motions
            print(f'[INFO] Skipping motion: {motion["text"]} from URL: {motions_url} because it has no rounds.')
            continue

        rounds_parsed = {} #for tracking multiple rounds for the same motion
        results = {}

        rounds = motion['rounds']
        for i,round in enumerate(rounds): 
            round_url = round['round']

            round_json = None
            try: 
                async with session.get(round_url) as round_response:
                    assert round_response.status == 200, f"Failed to fetch round for {round_url}: {round_response.status}"
                    round_json = await round_response.json()
            except Exception as e:
                print(f'[ERROR] Failed to fetch round for URL: {round_url}, {e}')
                continue

            rounds_parsed[f'round_{i+1}'] = round_json['name']
            rounds_parsed[f'stage_{i+1}'] = round_json['stage']

            pairings_url = round_json['_links']['pairing']

            if is_ballotable:
                results = await retrieve_results(pairings_url, session)

        row = {
            'Motion': motion['text'],
            'Info Slide': motion['info_slide_plain'],
            'Tournament': tournament_name,
            'short_name': tournament_short_name,
            'tab_url': tab_url,
            **results,
            **rounds_parsed,
        }

        rows.append(row)

    return rows
```

```python
def parse_motion(div): 
    """
    Given a motion top-level div, parse out the relevant details.
    """
    round = div.find('span', class_="badge").get_text(strip=True)
    motion = div.find('h4').find(string=True, recursive=False).strip()

    info_slide = div.find(string=re.compile("View Info Slide"))
    info_slide_text = None

    if info_slide is not None:
        info_slide_text = info_slide.find_next('div', class_='modal-body').get_text(strip=True)

    # regex match the wins for prop and opp
    text = div.get_text()

    re_prop = re.compile(r"(\d+) (Prop|Gov) win")
    re_opp = re.compile(r"(\d+) Opp win")

    prop_wins = re.findall(re_prop, text)[0][0]
    opp_wins = re.findall(re_opp, text)[0][0]

    balance = div.find(string=re.compile('balance')).get_text(strip=True)

    return motion, round, info_slide_text, prop_wins, opp_wins, balance
```

```python
data = {
    "Tournament": [],
    "Motion": [],
    "Round": [],
    "Info Slide": [],

    "Prop wins": [],
    "Opp wins": [],
    "Balance": [],
}

for subdomain, slug in calico_tabs:
    calico_url = f"https://{subdomain}.calicotab.com/{slug}"

    motions_suburl = f"motions/statistics"
    motions_url = f"{calico_url}/{motions_suburl}"

    response = requests.get(motions_url) 
    try: 
        assert response.status_code == 200, f"Failed to fetch {motions_url}: {response.status_code}"
    except: 
        continue

    html = response.text
    soup = BeautifulSoup(html, 'html.parser')

    try:
        motion_divs = soup.find_all('div', class_="list-group mt-3")
        print(f"Found {len(motion_divs)} motions from {slug}.")

        for start in tqdm(range(len(motion_divs))):
            div = motion_divs[start]
            motion, round, info_slide_text, prop_wins, opp_wins, balance = parse_motion(div)
            

            data["Motion"].append(motion)
            data["Round"].append(round)
            data["Info Slide"].append(info_slide_text)
            data["Prop wins"].append(int(prop_wins))
            data["Opp wins"].append(int(opp_wins))
            data["Balance"].append(balance)
            data["Tournament"].append(slug)
    except: 
        print(f'[WARN] could not parse motions for url: {calico_url}')
```

## Parse WSDC motions from .txt on GitHub
https://github.com/tokyodebate/motions/blob/main/International/WSDC.txt

```python
import requests

url = 'https://raw.githubusercontent.com/tokyodebate/motions/refs/heads/main/International/WSDC.txt'

response = requests.get(url)
wsdc_motions_raw = response.text
```

```python
def count_tabs(line: str):
    """
    Count the number of tabs in a line.
    """
    return line.count('\t')
```

```python
data2 = {
    "Tournament": [],
    "Motion": [],
    "Round": [],
    "Info Slide": [],

    # "Prop wins": [],
    # "Opp wins": [],
    # "Balance": [],
}

def parse_wsdc_motions():
    lines = wsdc_motions_raw.split('\n')
    lines = [line.strip('\n') for line in lines]
    lines = lines[1:] # Skip the header

    tournament = None
    round = None

    for i in range(len(lines)):
        curr = lines[i]

        if count_tabs(curr) == 1: 
            tournament = curr.strip()
        elif count_tabs(curr) == 2:
            round = curr.strip()
        elif count_tabs(curr) == 3:
            next_line = lines[i+1] if i+1 < len(lines) else None
            info_text = next_line.strip() if next_line and count_tabs(next_line) == 4 else None

            motion = curr.strip()   
            data2["Motion"].append(motion)
            data2["Round"].append(round)
            data2["Info Slide"].append(info_text)
            data2["Tournament"].append(tournament)

        else:
            continue

parse_wsdc_motions()
```
